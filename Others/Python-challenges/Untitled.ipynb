```python
L = [1]

def NumToList(n):
    r = n%10
    q = n//10
    n_L = [r]
    
    while q != 0:
        r = q%10
        q = q//10
        n_L.append(r)
    
    return n_L[::-1]

def ListToNum(L):
    n = len(L)
    s = 0
    for i in range(0,n):
        s += L[i]*(10**(n-i-1))
    return s


def next_n(n):
    L = NumToList(n)
    next_n = []
    c = 1
    for i in range(1, len(L)):
        if L[i] == L[i-1]:
            c += 1

        else:
            next_n += [c, L[i-1]]
            c = 1
    next_n += [c, L[-1]]    

    return ListToNum(next_n)

for i in range(30):
    L.append(next_n(L[-1]))
#len(NumToList(L[30]))
```
