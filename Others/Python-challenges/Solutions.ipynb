```python
#level 2

def crypt(message):
    trans = message.maketrans("ABCDEFGHIJKLMNOPQRSTUVWXYZ".lower(),
                          "CDEFGHIJKLMNOPQRSTUVWXYZAB".lower())

    return message.translate(trans)

print(f"Next url -> http://www.pythonchallenge.com/pc/def/{crypt('map')}.html")
```

```python
#level 3
from collections import Counter

word = ""

with open("text/level3.txt","r") as f:
    text = f.read() 
    letter = collections.Counter(text)
    
for i, j in letter.most_common():
    if j == 1:
        word += i

print(f"Next url -> http://www.pythonchallenge.com/pc/def/{word}.html")
```

```python
#level 4
import re 

def search(text):
    return re.findall(r"[a-z][A-Z]{3}([a-z])[A-Z]{3}[a-z]", text)

with open("text/level4.txt", "r") as f:
    text = f.read()
    match = search(text)

print(f"Next url -> http://www.pythonchallenge.com/pc/def/{''.join(match)}.php")
```
