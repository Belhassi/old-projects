```python
import requests
#from time import sleep
from os import path, getcwd
import pandas as pd
import requests
from requests_html import HTML
import sys
```

```python
FILE_PATH = getcwd()
DATA_DIR = path.join(FILE_PATH, 'Data')
RAW_DATA = path.join(DATA_DIR, 'Raw Data')
CLEANED_DATA_DIR = path.join(DATA_DIR, 'Cleaned-data')
CSV_FILE = path.join(RAW_DATA, 'Top anime P:1.csv')
```

```python
anime_data = pd.read_csv(CSV_FILE)
anime_title = anime_data['Title']
```

```python
def url_to_txt(url, filename="Top anime.html", save=False):
    r = requests.get(url)
    if r.status_code == 200:
        html_text = r.text
        if save:
            with open(f"Top.html", 'w') as f:
                f.write(html_text)
        return html_text
    print("Failed")
    sys.exit(1)
```

```python
Data = []
for title in anime_title:
    
    title_link = {}
    url = f'https://myanimelist.net/search/all?q={title}&cat=all'
    
    html_text = url_to_txt(url)
    html = HTML(html=html_text)
    TheClass = '.hoverinfo_trigger'
    Area =  html.find(TheClass)
    Area = Area[0].find('a')
    link = Area[0].links
    
    title_link[title] = link
    Data.append(title_link)
    print("Done")

print(Data)
```

```python
for _dict_ in Data: 
    for title,link in _dict.items():
        print("h")
```
