```python
import requests
from os import path, getcwd
import pandas as pd
import requests
from requests_html import HTML
import sys
import re
```

```python
FILE_PATH = getcwd()
DATA_DIR = path.join(FILE_PATH, 'Data')
RAW_DATA = path.join(DATA_DIR, 'Raw Data')
CLEANED_DATA_DIR = path.join(DATA_DIR, 'Cleaned-data')
CSV_FILE = path.join(RAW_DATA, 'Top anime P:1.csv')
outpt_file = path.join(CLEANED_DATA_DIR, 'Final Resultat.csv')
```

```python
anime_data = pd.read_csv(CSV_FILE)
df = pd.DataFrame(anime_data)
df['ID'] = '-'
df['Popularity'] = '-'
```

```python
def url_to_txt(url, filename="Top anime.html", save=False):
    r = requests.get(url)
    if r.status_code == 200:
        html_text = r.text
        if save:
            with open(f"Top.html", 'w') as f:
                f.write(html_text)
        return html_text
    print("Failed")
    sys.exit(1)
```

```python
def Get_link(Title):
    title = '_'.join(Title.split(" "))
    url = f'https://myanimelist.net/search/all?q={title}&cat=all'
    
    html_text = url_to_txt(url)
    html = HTML(html=html_text)
    TheClass = '.hoverinfo_trigger'
    Area =  html.find(TheClass)
    Area = Area[0].find('a')
    
    link = Area[0].links
    link = list(link)[0]
    
    return link
```

```python
def get_id_from_link(link):
    regex_pattren =r'https://myanimelist.net/anime/(?P<ID>[\w-]+)/(?P<Title>[\w]+)'
    regex = re.compile(regex_pattren)
    match = regex.match(link)
    if match != None:
        ID = match['ID'] 
        return ID
    return "the is No Id"
```

```python
def Add_ID_to_csv(row):
    link = Get_link(row['Title'])
    row['ID'] =get_id_from_link(link)
    return row
        
df = df.apply(Add_ID_to_csv, axis=1)
```
