```python
import pandas as pd
import os
```

```python
FILE_PATH = os.getcwd()
RAW_DATA = os.path.join(FILE_PATH, 'Data','Raw Data')
output = os.path.join(FILE_PATH, 'Data','Cleaned-data', 'Top-anime-cleaned.csv')
output_2x = os.path.join(FILE_PATH, 'Data', 'Cleaned-data', 'Top-anime-cleaned-2x.csv')
all_files = os.listdir(RAW_DATA)
```

```python
DataFrames = []
csv_files = [file for file in all_files if file.endswith(".csv")]
for file in csv_files:
    csv_path = os.path.join(RAW_DATA, file)
    this_df = pd.read_csv(csv_path)
    DataFrames.append(this_df)
```

```python
all_data = pd.concat(DataFrames)
all_data.sort_values(by=['Rank'], inplace=True)
```

```python
def from_ten_to_twenty(row):
    row['Score'] = row['Score']*2
    return row

Score_2x = all_data.apply(from_ten_to_twenty, axis=1)
all_data.head(n=3100)
```

```python
all_data.to_csv(output, index=False)
Score_2x.to_csv(output_2x, index=False)
```
