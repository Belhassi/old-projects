```python
import os
from bs4 import BeautifulSoup
import pandas as pd
```

```python
NOTE_DIR = os.getcwd()
BASE_DIR = os.path.dirname(NOTE_DIR)
HTML_DIR = os.path.join(BASE_DIR, "HTML")
DATA_DIR = os.path.join(BASE_DIR, "DATA")
RAW_DIR = os.path.join(DATA_DIR, "Raw")
os.makedirs(DATA_DIR, exist_ok=True)
```

```python
def open_html(fname):
    fname = str(fname)
    path = os.path.join(HTML_DIR, fname)+".html"
    
    with open(path, "r") as f:
        reader = f.read()
        return reader
```

```python
def scrape(page):
    All_data = []
    soup = BeautifulSoup(page, "lxml")
    Title = soup.title.string.split()[3]
    Contents = soup.find_all('div', class_="panel panel-aya aya_holder")
      
    for Content in Contents:
        Data = {}
        
        Aya = Content.find_all('div', class_="aya_text")
        Tafssir = Content.find_all('div', class_="trans_text")
        Aya_n = Content.find('div', class_="panel-heading")
                
        Data['َAya'] = Aya[0].text.replace('\n', '')
        Data['Tafssir'] = Tafssir[0].text
        Data['Aya Number'] = Aya_n.text.split()[1]
        Data['chapter'] = Title
        
        All_data.append(Data)
    return All_data, Title
```

```python
def save(Data, fname, n):
    path = os.path.join(RAW_DIR, f"{fname}|{n}.csv")
    df =  pd.DataFrame(Data)
    df.to_csv(path, index=False)
```

```python
files = os.listdir(HTML_DIR)
f = lambda file : int(file.replace(".html", ""))
files = map(f, files)

All_Data = []
for n_file in sorted(files):
    file = open_html(n_file)     
    Data, Title = scrape(file)
    
    save(Data, "سورة "+Title, n_file)
    print("Done with ", Title)
```
