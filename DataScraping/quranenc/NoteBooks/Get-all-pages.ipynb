```python
import requests
import os 
import time
```

```python
NOTE_DIR = os.getcwd()
BASE_DIR = os.path.dirname(NOTE_DIR)
HTML_DIR = os.path.join(BASE_DIR, "HTML")
os.makedirs(HTML_DIR, exist_ok=True)
```

```python
for pages_n in range(1, 110+1):
    url = f"https://quranenc.com/ar/browse/arabic_moyassar/{pages_n}"
    html_path = os.path.join(HTML_DIR, f"{pages_n}.html")
    
    html = requests.get(url)
    with open(html_path, "w") as f:
        f.write(html.text)
    print("Done with %s" %url)
    time.sleep(1)
```
